```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


sns.set_theme(style="whitegrid", palette="viridis")
plt.rcParams["figure.figsize"] = (10, 5)
pd.set_option("display.max_columns", None)


df = pd.read_csv("train.csv")
print(f"Wymiary zbioru: {df.shape[0]} wierszy, {df.shape[1]} kolumn")
df.head()
```

```python

display(df.describe())


print("\n--- Liczba kolumn według typu danych ---")
print(df.dtypes.value_counts())


na_summary = pd.DataFrame({
    "Liczba_brakow": df.isna().sum(),
    "Procent_brakow": (df.isna().sum() / len(df) * 100).round(2),
    "Typ_danych": df.dtypes
})
na_summary = na_summary[na_summary["Liczba_brakow"] > 0].sort_values(by="Liczba_brakow", ascending=False)

print("\n--- Kolumny z wartościami NaN ---")
display(na_summary)


plt.figure(figsize=(10, 6))
sns.barplot(x=na_summary["Procent_brakow"], y=na_summary.index, color="#d95f02")
plt.title("Odsetek wartości NaN po domyślnym pd.read_csv()")
plt.xlabel("% wierszy")
plt.ylabel("Kolumna")
plt.xlim(0, 105)
plt.show()
```

```python
sp = df["SalePrice"]


sp_min = sp.min()
sp_max = sp.max()
sp_mean = sp.mean()
sp_median = sp.median()
q1 = sp.quantile(0.25)
q3 = sp.quantile(0.75)
iqr = q3 - q1
upper_fence = q3 + 1.5 * iqr
outliers_count = (sp > upper_fence).sum()

print(f"Minimum:        {sp_min:,.0f} USD")
print(f"Maksimum:       {sp_max:,.0f} USD")
print(f"Średnia:        {sp_mean:,.2f} USD")
print(f"Mediana (Q2):   {sp_median:,.0f} USD")
print(f"Q1 (25%):       {q1:,.0f} USD")
print(f"Q3 (75%):       {q3:,.0f} USD")
print(f"IQR (Q3 - Q1):  {iqr:,.0f} USD")
print(f"Górny płot:     {upper_fence:,.0f} USD (liczba obserwacji powyżej: {outliers_count})")
print(f"Skośność:       {sp.skew():.2f}")


fig, axes = plt.subplots(1, 3, figsize=(18, 5))
for ax, b in zip(axes, [10, 30, 50]):
    sns.histplot(sp / 1000, bins=b, kde=True, ax=ax, color="#2b7a78")
    ax.axvline(sp_mean / 1000, color="orange", linestyle="--", linewidth=2, label=f"Średnia: {sp_mean/1000:.0f}k")
    ax.axvline(sp_median / 1000, color="black", linestyle="-", linewidth=2, label=f"Mediana: {sp_median/1000:.0f}k")
    ax.set_title(f"Histogram cen (bins = {b})")
    ax.set_xlabel("Cena sprzedaży (tys. USD)")
    ax.set_ylabel("Liczba domów")
    ax.legend()

plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 3))
sns.boxplot(x=sp / 1000, color="#69b3a2")
plt.title("Wykres pudełkowy zmiennej SalePrice")
plt.xlabel("Cena sprzedaży (tys. USD)")
plt.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(20, 5))


sns.boxplot(data=df, x="OverallQual", y=df["SalePrice"] / 1000, ax=axes[0], color="#8da0cb")
axes[0].set_title("Cena a ogólna jakość domu (1–10)")
axes[0].set_xlabel("Ogólna jakość (OverallQual)")
axes[0].set_ylabel("Cena sprzedaży (tys. USD)")


elec_counts = df["Electrical"].value_counts()
sns.barplot(x=elec_counts.values, y=elec_counts.index, ax=axes[1], color="#2b7a78")
axes[1].set_title("Liczba domów wg instalacji elektrycznej")
axes[1].set_xlabel("Liczba domów")
axes[1].set_ylabel("Typ instalacji")


sales_time = df.groupby(["YrSold", "MoSold"]).size().reset_index(name="Count")
sales_time["Date"] = pd.to_datetime(sales_time["YrSold"].astype(str) + "-" + sales_time["MoSold"].astype(str) + "-01")
sns.lineplot(data=sales_time, x="Date", y="Count", marker="o", ax=axes[2], color="#2b7a78")
axes[2].set_title("Sprzedaż domów według miesiąca (2006–2010)")
axes[2].set_xlabel("Data sprzedaży")
axes[2].set_ylabel("Liczba sprzedanych domów")

plt.tight_layout()
plt.show()
```

```python

area_cols = [col for col in df.columns if "Area" in col or "SF" in col]
print("Zmienne związane z powierzchnią:")
print(area_cols)


corrs = df[area_cols + ["SalePrice"]].corr()["SalePrice"].sort_values(ascending=False)
print("\nKorelacja Pearsona z SalePrice:")
print(corrs)


plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x="GrLivArea", y=df["SalePrice"] / 1000, hue="SaleCondition", alpha=0.7)
plt.axvline(4000, color="red", linestyle="--", label="Próg 4000 stóp²")
plt.title("Powierzchnia mieszkalna (GrLivArea) a cena sprzedaży")
plt.xlabel("Powierzchnia mieszkalna nadziemna [stopy²]")
plt.ylabel("Cena sprzedaży [tys. USD]")
plt.legend(title="Warunki sprzedaży", bbox_to_anchor=(1.05, 1), loc="upper left")
plt.tight_layout()
plt.show()


print("\nObserwacje odstające (GrLivArea > 4000 stóp²):")
display(df[df["GrLivArea"] > 4000][["Id", "GrLivArea", "SalePrice", "OverallQual", "SaleCondition", "Neighborhood"]])


selected_areas = ["GrLivArea", "TotalBsmtSF", "GarageArea", "LotArea"]
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

for ax, col in zip(axes.flatten(), selected_areas):
    r = df[col].corr(df["SalePrice"])
    sns.regplot(
        data=df, x=col, y=df["SalePrice"] / 1000, ax=ax,
        scatter_kws={"alpha": 0.4, "color": "#2b7a78"},
        line_kws={"color": "#d95f02"}
    )
    ax.set_title(f"{col} vs SalePrice (r = {r:.2f})")
    ax.set_xlabel(f"{col} [stopy²]")
    ax.set_ylabel("Cena sprzedaży [tys. USD]")

plt.tight_layout()
plt.show()
```

Podsumowanie odpowiedzi na pytania z zadań

1. Struktura zbioru i braki danych
Zmienna docelowa (target): SalePrice (cena sprzedaży nieruchomości w USD).
Rodzaje zmiennych: Zbiór liczy 1460 wierszy i 81 kolumn, w tym 38 kolumn liczbowych (int64, float64, obejmujących m.in. metraż, lata, oceny jakości od 1 do 10) oraz 43 kolumny tekstowe/kategoryczne (object, np. dzielnica, typ budynku, wykończenie).
Pułapka z brakami (NaN): Po wczytaniu pliku w 19 kolumnach pojawiają się braki NaN, jednak zgodnie z dokumentacją data_description.txt w 14 z nich tekst "NA" nie oznacza błędu pomiaru, lecz brak danego udogodnienia (np. PoolQC to brak basenu w 99,5% domów, Alley to brak alejki, Fence to brak ogrodzenia, FireplaceQu to brak kominka, kolumny Garage to brak garażu, kolumny Bsmt to brak piwnicy).
Rzeczywiste braki danych: Występują przede wszystkim w LotFrontage (259 braków, ok. 17,7%), MasVnrArea (8 braków) oraz Electrical (1 brak).

2. Analiza zmiennej docelowej SalePrice
Statystyki podstawowe:
Minimum: 34 900 USD | Maksimum: 755 000 USD
Średnia: 180 921,20 USD | Mediana (Q2): 163 000 USD
Q1 (25%): 129 975 USD | Q3 (75%): 214 000 USD | IQR: 84 025 USD
Rozkład i skośność: Ceny nie mają rozkładu normalnego. Histogram jest wyraźnie prawostronnie skośny (współczynnik skośności wynosi 1,88), przez co średnia (181 tys. USD) jest wyższa od mediany (163 tys. USD), ponieważ zawyża ją długi ogon najdroższych rezydencji.
Wartości odstające: Na wykresie pudełkowym widać 61 obserwacji powyżej górnego płotu (Q3 + 1,5 * IQR = 340 037,5 USD). Oznaczają one po prostu luksusowe, duże domy o wysokim standardzie (OverallQual od 8 do 10) w prestiżowych dzielnicach, więc nie są błędami i nie należy ich automatycznie usuwać.

3. Wpływ powierzchni na cenę domu
Hipoteza („Większe domy są droższe”): Potwierdzona. Powierzchnia mieszkalna nadziemna (GrLivArea) wykazuje silną, dodatnią i w przybliżeniu liniową zależność z ceną (r = 0,71).
Nietypowe obserwacje (anomalie): Powyżej 4000 stóp kwadratowych znajdują się 4 domy:
Dwa z nich (ID 692 oraz 1183) kosztowały ok. 750 tys. USD, co pasuje do ogólnego trendu.
Dwa pozostałe (ID 524 o pow. 4676 stóp kw. za 184 750 USD oraz ID 1299 o pow. 5642 stóp kw. za 160 000 USD) mają najwyższą ocenę jakości (10), ale zostały sprzedane bardzo tanio jako sprzedaż częściowa (SaleCondition = Partial), przez co nie odzwierciedlają normalnej ceny rynkowej i zaburzają liniowość modelu.
Porównanie rodzajów powierzchni (korelacja Pearsona z SalePrice):
1. GrLivArea (powierzchnia mieszkalna nadziemna): najsilniejszy związek z ceną (r = 0,71).
2. GarageArea (powierzchnia garażu): umiarkowanie silny związek (r = 0,62).
3. TotalBsmtSF (powierzchnia piwnicy): umiarkowanie silny związek (r = 0,61).
4. LotArea (powierzchnia działki): znacznie słabszy związek (r = 0,26); duża działka (np. na obrzeżach miasta) nie gwarantuje wysokiej ceny, jeśli sam budynek jest mały lub w słabym stanie.
